```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from scipy import stats
from sklearn.preprocessing import MinMaxScaler
```

```python
raw_data = pd.read_excel('weather-sa-2017-2019-v4.xlsx')
```

```python
raw_data.head()

```

```python

raw_data = raw_data.drop_duplicates()
raw_data = raw_data.replace('ERR', np.nan)
raw_data = raw_data.replace(-1, np.nan)
raw_data.info()
```

```python
imputer = SimpleImputer(missing_values = np.nan,strategy = 'mean')
raw_data[['temp','wind','humidity','barometer','visibility']]=imputer.fit_transform(raw_data[['temp','wind','humidity','barometer','visibility']])
raw_data
```

```python

raw_data.dropna(inplace=True)
raw_data.info()
```

```python
raw_data['city']=raw_data['city'].astype(str)
raw_data['month']=raw_data['month'].astype(int)
raw_data['day']=raw_data['day'].astype(int)
raw_data.info()
```

```python
print(raw_data['city'].unique())
```

```python
raw_data['city']=raw_data['city'].replace('EP','Eastern Province')
raw_data['city']=raw_data['city'].replace('Baha','Al-baha')
print(raw_data['city'].unique())
```

```python
raw_data.head()
```

```python
raw_data['rain'] = raw_data['rain'].astype(bool).map({False: 0, True: 1})
raw_data.head()
```

```python

numeric_columns = raw_data.select_dtypes(include=['float64', 'int64']).columns

for column in numeric_columns:
    Q1 = raw_data[column].quantile(0.25)
    Q3 = raw_data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    raw_data = raw_data[(raw_data[column] >= lower_bound) & (raw_data[column] <= upper_bound)]
```

```python
def time_of_day(time):
    if 5 <= time < 12:
        return 'Morning'
    elif 12 <= time < 18:
        return 'Afternoon'
    elif 18 <= time < 24:
        return 'Evening'
    else:
        return 'Night'

raw_data['time'] = raw_data['hour'].apply(lambda x: time_of_day(x))
raw_data.head()
```

```python
def wind_category(windSpeed):
    if windSpeed < 10:
        return 'Low'
    elif 10 <= windSpeed < 20:
        return 'Medium'
    else:
        return 'High'

raw_data['windSpeed'] = raw_data['wind'].apply(lambda x: wind_category(x))
raw_data.head()
```

```python
raw_data['time'] = raw_data['time'].astype(str).map({'Morning': 0, 'Afternoon': 1,'Evening': 2,'Night':3 })

raw_data.head()
```

```python
raw_data['windSpeed'] = raw_data['windSpeed'].astype(str).map({'Low': 0, 'Medium': 1,'High': 2 })
raw_data.head()
```

```python
raw_data = pd.get_dummies(raw_data, columns=['city']).astype('int64')
raw_data.head()
```

```python
raw_data['date']= pd.to_datetime(raw_data['date'], unit='ns')
raw_data

```

```python
raw_data.drop(["date","rain","minute"],inplace=True,axis=1)
raw_data.head()
```

```python
raw_data.info()
```

```python
scaler = MinMaxScaler()
ml_data = scaler.fit_transform(raw_data)
ml_data
```
